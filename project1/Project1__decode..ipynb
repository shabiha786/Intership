```python
import pandas as pd

df = pd.read_excel("Dataset for Data Analytics.xlsx")

# 1. Missing values check
print(df.isnull().sum())

# 2. Handle missing CouponCode (categorical -> label, not deletion)
df['CouponCode'] = df['CouponCode'].fillna("No Coupon")

# 3. Duplicate check (unique identifier audit)
dupe_ids = df['OrderID'].duplicated().sum()
full_dupes = df.duplicated().sum()
print("Duplicate OrderIDs:", dupe_ids, "| Full duplicate rows:", full_dupes)
df = df.drop_duplicates()

# 4. Date format standardize (ISO 8601)
df['Date'] = pd.to_datetime(df['Date']).dt.strftime('%Y-%m-%d')

# 5. Text cleanup (trim + consistent case) - safety net even if already clean
text_cols = ['Product','ShippingAddress','PaymentMethod','OrderStatus','ReferralSource']
for col in text_cols:
    df[col] = df[col].str.strip()

# 6. Numeric precision check
df['UnitPrice'] = df['UnitPrice'].round(2)
df['TotalPrice'] = df['TotalPrice'].round(2)

df.to_excel("Dataset_Cleaned.xlsx", index=False)
```
